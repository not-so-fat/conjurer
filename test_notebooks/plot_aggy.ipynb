```python
# Timestamp X — daily cost (with a gap day)
rng = pandas.date_range("2024-01-01", periods=8, freq="D")
# skip Jan 4 intentionally
days = [d for d in rng if d.day != 4]
df_ts = pandas.DataFrame({
    "timestamp": pandas.to_datetime(
        [d + pandas.Timedelta(hours=int(random.randint(0, 23))) for d in days for _ in range(3)]
    ),
    "cost": random.uniform(5, 50, len(days) * 3),
    "txn_id": range(len(days) * 3),
})
df_ts.head()
```

```python
eda.plot_aggy(df_ts, "timestamp", "cost", agg="sum", freq="D")  # fill_empty default True
```

```python
eda.plot_aggy(df_ts, "timestamp", "cost", agg="sum", freq="D", fill_empty=False)
```

```python
eda.plot_aggy(df_ts, "timestamp", "txn_id", agg="count", freq="D")
```

```python
eda.plot_aggy(df_ts, "timestamp", "cost", agg="mean", freq="D", mark="line")
```
